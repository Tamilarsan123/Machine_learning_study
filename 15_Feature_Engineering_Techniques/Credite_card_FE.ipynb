## 1.Import Necessary Libraties

```python
import pandas as pd 

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.feature_selection import RFE

from sklearn.metrics import accuracy_score,precision_score,recall_score,confusion_matrix

import warnings 
warnings.filterwarnings("ignore")
```

## 2.Import Dataset

```python
credit_card_data = pd.read_csv("credit_card_clean.csv")
credit_card_data
```

## 4.Data Preppuration

### 4.1 Remove Unwanted columns

```python
credit_card_data.head()
```

```python
del credit_card_data["ID"]
```

```python
credit_card_data.head()
```

### ID is unwanted for the model prediction

### 4.2 change the Data type

```python
le_model = LabelEncoder()
credit_card_data["SEX"] = le_model.fit_transform(credit_card_data["SEX"])
credit_card_data["EDUCATION"] = le_model.fit_transform(credit_card_data["EDUCATION"])
credit_card_data["MARRIAGE"] = le_model.fit_transform(credit_card_data["MARRIAGE"])
credit_card_data.head()
```

```python
x = credit_card_data.drop(labels="DEFAULT",axis=1)
y = credit_card_data.DEFAULT
```

#### Change Input value scale

```python
sd_scale = StandardScaler()
scaled_x = sd_scale.fit_transform(x)
```

## 5.Model Building

```python
x_train,x_test,y_train,y_test = train_test_split(scaled_x,y,test_size=0.20,shuffle=True,stratify=y)
```

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

## 6.Model Training || 7.Model Testing || 8.Model Evaluation

```python
def runrandomforestclassifier(x_train,x_test,y_train,y_test):
    rd_model = RandomForestClassifier(max_depth=10)
    rd_model.fit(x_train,y_train)

    y_pred_test = rd_model.predict(x_test)
    print("Accuracy_score  :",round(accuracy_score(y_test,y_pred_test),4))
    print("Precision_score :",round(precision_score(y_test,y_pred_test),4))
    print("Recall_score    :",round(recall_score(y_test,y_pred_test),4))
    print("Confusion_matrix:\n",confusion_matrix(y_test,y_pred_test))
```

### REF Feature selection

```python
for i in range(1,25):
    ref_model = RFE(estimator = GradientBoostingClassifier(),n_features_to_select=i)
    ref_model.fit(x_train,y_train)

    x_train_ref_model = ref_model.transform(x_train)
    x_test_ref_model = ref_model.transform(x_test)

    print("Feature selection:",i)
    runrandomforestclassifier(x_train_ref_model,x_test_ref_model,y_train,y_test)
    print("**************************")
```
