```python
breast_cancer_data = load_breast_cancer()
```

```python
breast_cancer_data_df = pd.DataFrame(data = breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_data_df["target"] = breast_cancer_data.target
breast_cancer_data_df
```

# 4.Model Building

```python
x = breast_cancer_data_df.drop(labels = ["target"],axis=1)
y = breast_cancer_data_df["target"]
```

```python
x.shape,y.shape
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size =0.20,random_state=78,shuffle=True,stratify=y)
#startify its mean the Y_test samle not get a equal data point at malignent cancer and benign canser so use to fix the equal leval of data point fix
```

```python
print("Taining_samples",x_train.shape,y_train.shape)
print("Testing_samples",x_test.shape,y_test.shape)
```

```python
%%time
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(criterion='gini',max_depth=5,class_weight={0:2,1:1})
# max_depth --> this how much level dt growth con fix
# class weight ---> one o/p feature only get 37% so focuse double more so use class weight fix equal priority for two class output feature
dt_model.fit(x_train,y_train)
```

# 6.Model Testing || 7. Model Evaluation

## Training Data

```python
y_pred_train = dt_model.predict(x_train)
y_pred_train
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
confusion_matrix(y_train,y_pred_train)
```

```python
print(classification_report(y_train,y_pred_train))
```

## Test Data

```python
y_pred_test = dt_model.predict(x_test)
y_pred_test
```

```python
confusion_matrix(y_test,y_pred_test)
```

```python
print(classification_report(y_test,y_pred_test))
```
